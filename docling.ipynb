```python
# Required Libraries

from docling.document_converter import DocumentConverter
from docling.datamodel.base_models import InputFormat
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.backend.pypdfium2_backend import PyPdfiumDocumentBackend
import json

print("Imports successful!")
```

```python
# Define the Path of PDF

PDF_PATH = "HDFC Income Fund.pdf"
```

```python
# Initialize the DocumentConverter

converter = DocumentConverter()

print("DocumentConverter Initialised Succesfully")
```

```python
# Convert the PDF Document

result = converter.convert(PDF_PATH)

doc = result.document

print("Document Converted Succefully")
print(f"Number of Pages: {doc.num_pages}")
```

```python
# Document Elements in Reading Order

print("=" * 80)
print("Document Elements in Reading Order")
print("=" * 80)

for idx, (item, level) in enumerate(doc.iterate_items()):

    item_type = type(item).__name__

    text_preview = ""
    if hasattr(item, 'text'):
        text_preview = item.text[:100] + "..." if len(item.text) > 100 else item.text

    print(f"\n[{idx+1}] Type: {item_type} | Level: {level}")
    if text_preview:
        print(f" Text: {text_preview}")


    if idx >=19:
        print("\n... (showing first 20 items)")
        break
```

```python
# Layout Information with Bounding Boxes

print("=" * 80)
print("Layout Information (Bounding Boxes)")
print("=" * 80)

for idx, (item, level) in enumerate(doc.iterate_items()):
    item_type = type(item).__name__

    if hasattr(item, 'prov') and item.prov:
        for prov in item.prov:
            page_no = prov.page_no if hasattr(prov, 'page_no') else 'N/A'


            bbox = prov.bbox if hasattr(prov, 'bbox') else None

            print(f"\n[{idx + 1}] {item_type}")
            print(f" page: {page_no}")
            if bbox:
                print(f"Bounding Box")
                print(f"  - Left: {bbox.l:.2f}")
                print(f"  - Right: {bbox.r:.2f}")
                print(f"  - Top: {bbox.t:.2f}")
                print(f"  - Bottom: {bbox.b:.2f}")

        if idx >= 9:
            print("\n...(Showing the first 10 items with layout)")
            break
```

```python
# Table Extraction

from docling.datamodel.document import TableItem

print("=" * 80)
print("Table Extraction with Structure")
print("=" * 80)

table_count = 0

for idx, (item,level) in enumerate(doc.iterate_items()):

    if isinstance(item, TableItem):
        table_count +=1
        print(f"\n{'='*60}")
        print(f"TABLE {table_count}")
        print(f"{'='*60}")

        table_data = item.data

        if table_data:

            num_rows = table_data.num_rows if hasattr(table_data, 'num_rows') else 'N/A'
            num_cols = table_data.num_cols if hasattr(table_data, 'num_cols') else 'N/A'

            print(f"Dimensions: {num_rows} rows x {num_cols} columns")

            if hasattr(table_data, 'table_cells') and table_data.table_cells:
                print(f"\nCell Details (Showing first 10 cells:)")
                print("-" * 50)

                for cell_idx, cell in enumerate(table_data.table_cells[:10]):

                    row_span = getattr(cell, 'row_span', 'N/A')
                    col_span = getattr(cell, 'col_span', 'N/A')

                    row_idx = row_span[0] if isinstance(row_span, (list, tuple)) else row_span
                    col_idx = col_span[0] if isinstance(col_span, (list, tuple)) else col_span

                    cell_text = cell.text if  hasattr(cell, 'text') else''
                    cell_text = cell_text[:50] + '...' if len(cell_text) > 50 else cell_text

                    is_header = cell.is_header if hasattr(cell, 'is_header') else False
                

                    print(f"  Cell[Row:{row_idx}, Col:{col_idx}]" + 
                          (" [HEADER]" if is_header else "") +
                          f" = '{cell_text}'")



        if table_count >=3:
            print("\n...(showing first 3 tables)")
            break

print(f"\n\nTotal Tables Found in document: {table_count}")



```

```python
# Export Data to Pandas Dataframe

import pandas as pd

print("=" * 80)
print("Table Export To Pandas")
print("=" * 80)

table_count = 0

for idx, (item,level) in enumerate(doc.iterate_items()):
    if isinstance(item, TableItem):
        table_count += 1



        try:

            df = item.export_to_dataframe()
            print(f"\n Table {table_count} as DataFrame:")
            print(f"Shape: {df.shape}") #(rows, columns)
            print(df.head(10))

        except Exception as e:
            print(f"Could not export table {table_count}: {e}")

        if table_count >= 2:
            break
```

```python
# Export to Markdown


markdown_output = doc.export_to_markdown()

print("=" * 80)
print("Markdown Export (first 2000 Characters)")
print("=" * 80)
print(markdown_output[:2000])
```

```python
# Export to JSON

json_output = doc.export_to_dict()

print("=" * 80)
print("Json Export Structure")
print("=" * 80)
print(f"Top-level Keys: {list(json_output.keys())}")

if 'body' in json_output:
    print(f"\nNumber of body elements:{len(json_output['body'])}")
```
